```python
import boto3
import os

bedrock_control = boto3.client(
    service_name="bedrock",
    region_name="ap-southeast-6",
    aws_access_key_id=os.environ["AWS_ACCESS_KEY_ID"],
    aws_secret_access_key=os.environ["AWS_SECRET_ACCESS_KEY"]
)

response = bedrock_control.list_inference_profiles(
    typeEquals="SYSTEM_DEFINED"
)

print("Available inference profiles:\n")

for profile in response.get("inferenceProfileSummaries", []):
    print("Name:", profile["inferenceProfileName"])
    print("ID:", profile["inferenceProfileId"])
    print("ARN:", profile["inferenceProfileArn"])
    print("Status:", profile["status"])
    print("-" * 60)
```

```python
response = bedrock.converse(
    modelId="au.anthropic.claude-sonnet-4-6",
    messages=[
        {
            "role": "user",
            "content": [
                {
                    "text": "Hello! Please introduce yourself in one sentence."
                }
            ]
        }
    ],
    inferenceConfig={
        "maxTokens": 200,
        "temperature": 0.7
    }
)

claude_reply = response["output"]["message"]["content"][0]["text"]

print("Claude response:")
print(claude_reply)
```

```python
import fitz
import pandas as pd
import os
```

```python
pdf_files = [
    file_name
    for file_name in os.listdir("/content")
    if file_name.lower().endswith(".pdf")
]

print("PDF files found:", len(pdf_files))

for file_name in pdf_files:
    print("-", file_name)
```

```python
extracted_pages = []

for file_name in pdf_files:
    file_path = os.path.join("/content", file_name)

    with fitz.open(file_path) as pdf_document:
        for page_number, page in enumerate(pdf_document, start=1):
            page_text = page.get_text("text").strip()

            extracted_pages.append(
                {
                    "source_file": file_name,
                    "page_number": page_number,
                    "text": page_text,
                    "character_count": len(page_text)
                }
            )

print("Total pages extracted:", len(extracted_pages))
```

```python
pages_df = pd.DataFrame(extracted_pages)

pages_df.head()
```

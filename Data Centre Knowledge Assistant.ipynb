```python
def create_chunks(text, chunk_size=700, overlap=100):
    chunks = []
    start = 0

    while start < len(text):
        end = start + chunk_size
        chunk = text[start:end]

        chunks.append(chunk)

        start += chunk_size - overlap

    return chunks
```

```python
chunk_records = []

for _, row in pages_df.iterrows():

    chunks = create_chunks(row["clean_text"])

    for i, chunk in enumerate(chunks):

        chunk_records.append({
            "source_file": row["source_file"],
            "page_number": row["page_number"],
            "chunk_number": i + 1,
            "chunk_text": chunk
        })

chunks_df = pd.DataFrame(chunk_records)

print("Total Chunks:", len(chunks_df))

chunks_df.head()
```

```python
embeddings = model.encode(
    chunks_df["chunk_text"].tolist(),
    show_progress_bar=True,
    convert_to_numpy=True
)

print("Embedding Shape:", embeddings.shape)
```

```python
dimension = embeddings.shape[1]

index = faiss.IndexFlatL2(dimension)

index.add(embeddings)

print("FAISS Index Created Successfully!")
print("Total Vectors:", index.ntotal)
```

```python
def search_documents(query, top_k=3):
    # Convert the user query into an embedding
    query_embedding = model.encode(
        [query],
        convert_to_numpy=True
    )

    # Search the FAISS index
    distances, indices = index.search(query_embedding, top_k)

    results = []

    for i in indices[0]:
        results.append(chunks_df.iloc[i])

    return pd.DataFrame(results)
```

```python
def retrieve_context(query, top_k=3):
    query_embedding = model.encode(
        [query],
        convert_to_numpy=True
    )

    distances, indices = index.search(query_embedding, top_k)

    context = ""

    for idx in indices[0]:
        context += chunks_df.iloc[idx]["chunk_text"] + "\n\n"

    return context
```

```python
%%writefile app.py

import os
import glob
import re

import boto3
import faiss
import fitz
import numpy as np
import pandas as pd
import streamlit as st
from sentence_transformers import SentenceTransformer


# -----------------------------------------------------
# Page configuration
# -----------------------------------------------------

st.set_page_config(
    page_title="Data Centre Knowledge Assistant",
    page_icon="🏢",
    layout="wide"
)

st.title("🏢 Data Centre Knowledge Assistant")

st.write(
    "Ask questions about data centre procedures, incidents, "
    "maintenance, safety, handovers, and operational SOPs."
)


# -----------------------------------------------------
# Configuration
# -----------------------------------------------------

AWS_REGION = os.getenv("AWS_DEFAULT_REGION", "ap-southeast-6")
MODEL_ID = "au.anthropic.claude-sonnet-4-6"


# -----------------------------------------------------
# Text cleaning and chunking
# -----------------------------------------------------

def clean_text(text):
    text = text.replace("\n", " ")
    text = re.sub(r"\s+", " ", text)
    return text.strip()


def create_chunks(text, chunk_size=700, overlap=100):
    chunks = []
    start = 0
    step = chunk_size - overlap

    while start < len(text):
        chunk = text[start:start + chunk_size].strip()

        if chunk:
            chunks.append(chunk)

        start += step

    return chunks


# -----------------------------------------------------
# Load documents and build FAISS index
# -----------------------------------------------------

@st.cache_resource
def load_knowledge_base():
    pdf_files = glob.glob("/content/*.pdf")

    if not pdf_files:
        raise FileNotFoundError(
            "No PDF files were found in /content. "
            "Upload the SOP PDFs to Colab first."
        )

    chunk_records = []

    for file_path in pdf_files:
        source_file = os.path.basename(file_path)

        with fitz.open(file_path) as pdf_document:
            for page_number, page in enumerate(pdf_document, start=1):
                page_text = clean_text(page.get_text("text"))

                for chunk_number, chunk in enumerate(
                    create_chunks(page_text),
                    start=1
                ):
                    chunk_records.append({
                        "source_file": source_file,
                        "page_number": page_number,
                        "chunk_number": chunk_number,
                        "chunk_text": chunk
                    })

    chunks_df = pd.DataFrame(chunk_records)

    embedding_model = SentenceTransformer("all-MiniLM-L6-v2")

    embeddings = embedding_model.encode(
        chunks_df["chunk_text"].tolist(),
        convert_to_numpy=True,
        show_progress_bar=False
    ).astype("float32")

    dimension = embeddings.shape[1]

    faiss_index = faiss.IndexFlatL2(dimension)
    faiss_index.add(embeddings)

    return chunks_df, embedding_model, faiss_index


# -----------------------------------------------------
# Amazon Bedrock client
# -----------------------------------------------------

@st.cache_resource
def create_bedrock_client():
    return boto3.client(
        service_name="bedrock-runtime",
        region_name=AWS_REGION
    )


# -----------------------------------------------------
# Document retrieval
# -----------------------------------------------------

def retrieve_documents(
    question,
    chunks_df,
    embedding_model,
    faiss_index,
    top_k=5
):
    query_embedding = embedding_model.encode(
        [question],
        convert_to_numpy=True
    ).astype("float32")

    distances, indices = faiss_index.search(
        query_embedding,
        min(top_k, len(chunks_df))
    )

    retrieved_rows = []

    for position, chunk_index in enumerate(indices[0]):
        if chunk_index < 0:
            continue

        row = chunks_df.iloc[chunk_index]

        retrieved_rows.append({
            "source_file": row["source_file"],
            "page_number": int(row["page_number"]),
            "chunk_number": int(row["chunk_number"]),
            "chunk_text": row["chunk_text"],
            "distance": float(distances[0][position])
        })

    return retrieved_rows


# -----------------------------------------------------
# Claude RAG response
# -----------------------------------------------------

def ask_claude(question, retrieved_documents, bedrock_client):
    context_sections = []

    for document in retrieved_documents:
        context_sections.append(
            f"""
Source: {document["source_file"]}
Page: {document["page_number"]}
Content:
{document["chunk_text"]}
"""
        )

    context = "\n\n".join(context_sections)

    prompt = f"""
You are an AI Data Centre Operations Knowledge Assistant.

Answer the user's question using only the SOP context below.

Instructions:
- Do not invent procedures.
- Give clear and practical operational steps.
- Mention the relevant source document and page where possible.
- If the context does not contain enough information, say:
  "I could not find enough information in the available SOP documents."

SOP CONTEXT:
{context}

USER QUESTION:
{question}
"""

    response = bedrock_client.converse(
        modelId=MODEL_ID,
        messages=[
            {
                "role": "user",
                "content": [
                    {
                        "text": prompt
                    }
                ]
            }
        ],
        inferenceConfig={
            "maxTokens": 1000,
            "temperature": 0.1
        }
    )

    return response["output"]["message"]["content"][0]["text"]


# -----------------------------------------------------
# Load application resources
# -----------------------------------------------------

try:
    chunks_df, embedding_model, faiss_index = load_knowledge_base()
    bedrock_client = create_bedrock_client()

    st.success(
        f"Knowledge base ready: "
        f"{len(chunks_df)} searchable SOP chunks loaded."
    )

except Exception as error:
    st.error(f"Application setup failed: {error}")
    st.stop()


# -----------------------------------------------------
# User interface
# -----------------------------------------------------

question = st.text_input(
    "Ask a question",
    placeholder="Example: What should I do during a power outage?"
)

ask_button = st.button(
    "Ask the Assistant",
    type="primary"
)

if ask_button:
    if not question.strip():
        st.warning("Please enter a question.")

    else:
        with st.spinner("Searching SOP documents and consulting Claude..."):
            try:
                retrieved_documents = retrieve_documents(
                    question=question,
                    chunks_df=chunks_df,
                    embedding_model=embedding_model,
                    faiss_index=faiss_index,
                    top_k=5
                )

                answer = ask_claude(
                    question=question,
                    retrieved_documents=retrieved_documents,
                    bedrock_client=bedrock_client
                )

                st.subheader("Assistant response")
                st.markdown(answer)

                with st.expander("View retrieved SOP sources"):
                    for number, document in enumerate(
                        retrieved_documents,
                        start=1
                    ):
                        st.markdown(
                            f"""
**Result {number}**

- **Document:** {document["source_file"]}
- **Page:** {document["page_number"]}
- **Chunk:** {document["chunk_number"]}
"""
                        )

                        st.write(document["chunk_text"])
                        st.divider()

            except Exception as error:
                st.error(f"Unable to generate an answer: {error}")


# -----------------------------------------------------
# Footer
# -----------------------------------------------------

st.caption(
    "Built with Amazon Bedrock, Anthropic Claude, "
    "Sentence Transformers, FAISS, and Streamlit."
)
```
